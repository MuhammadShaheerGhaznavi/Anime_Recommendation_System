```python
import numpy  as np
import pandas as pd
```

```python
df = pd.read_csv("./dataset.csv/anime.csv")
synopsis = pd.read_csv("./dataset.csv/anime_with_synopsis.csv")
```

```python
#cleaning the data
df.dropna(inplace=True) 
synopsis.dropna(inplace=True)
synopsis =synopsis[['Name','sypnopsis']]
```

```python
df = df.merge(synopsis, on='Name')
```

```python
df = df[['MAL_ID','Name','Genres','sypnopsis','Type','Studios','Rating']]
```

```python
df
```

```python
# Ik animes without synopsis and genre data are small therefore drop 'em
df = df[~((df['sypnopsis'].str.contains("No synopsis information has been added to this")) & (df['Genres'] == 'Unknown'))]
```

```python
no_sypn = df[df['sypnopsis'].str.contains("No synopsis information has been added to this")]
no_sypn['sypnopsis'][2225]
```

```python
df_clean = df.replace('Unknown', np.nan)
df_clean = df_clean.replace('No synopsis information has been added to this title. Help improve our database by adding a synopsis here .',np.nan)
```

```python
no_sypn = df_clean[df_clean['sypnopsis'].isna()]
no_sypn
```

```python
df_clean['Genres'] = df_clean['Genres'].fillna('')
df_clean['sypnopsis'] = df_clean['sypnopsis'].fillna('')
df_clean['Rating'] = df_clean['Rating'].fillna('')
df_clean['Type'] = df_clean['Type'].fillna('')
df_clean['Studios'] = df_clean['Studios'].fillna('')
```

```python
df_clean['Rating'] = df_clean['Rating'].replace('R - 17+ (violence & profanity)', '17+')
df_clean['Rating'] = df_clean['Rating'].replace('PG-13 - Teens 13 or older', 'PG-13')
df_clean['Rating'] = df_clean['Rating'].replace('PG - Children', 'PG')
df_clean['Rating'] = df_clean['Rating'].replace('G - All Ages', 'G')
df_clean['Rating'] = df_clean['Rating'].replace('R+ - Mild Nudity', 'R+')
```

```python
df_clean['Genres'] = df_clean['Genres'].apply(lambda x:x.split(','))
df_clean['sypnopsis'] = df_clean['sypnopsis'].apply(lambda x:x.split())
df_clean['Rating'] = df_clean['Rating'].apply(lambda x:x.split())
df_clean['Type'] = df_clean['Type'].apply(lambda x:x.split())
df_clean['Studios'] = df_clean['Studios'].apply(lambda x:x.split())
```

```python
df_clean['Genres'] = df_clean['Genres'].apply(lambda x: [i.replace(" ","") for i in x] )
df_clean['sypnopsis'] = df_clean['sypnopsis'].apply(lambda x: [i.replace(" ","") for i in x] )
df_clean['Rating'] = df_clean['Rating'].apply(lambda x: [i.replace(" ","") for i in x] )
df_clean['Type'] = df_clean['Type'].apply(lambda x: [i.replace(" ","") for i in x] )
df_clean['Studios'] = df_clean['Studios'].apply(lambda x: [i.replace(" ","") for i in x] )
```

```python
df_clean
```

```python
df = df_clean
df['tags'] = df['sypnopsis'] +df['Genres'] + df['Rating']+ df['Studios']+df['Type']
```

```python
df =df[['MAL_ID', 'Name', 'tags']]
```

```python
df
```

```python
df['tags'] = df['tags'].apply(lambda x: ' '.join(x) )
df['tags'] =df['tags'].apply(lambda x: x.lower())
```

```python
df['tags'][0]
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
cv = CountVectorizer(max_features =5000, stop_words ='english')
```

```python
vector = cv.fit_transform(df['tags']).toarray()
vector.shape
```

```python
similarity = cosine_similarity(vector)
similarity
```

```python
df[df['Name']== 'Trigun'].index
```

```python
def recommend(anime):
    index = df[df['Name'] ==anime].index[0]
    distances = sorted(list(enumerate(similarity[index])),reverse=True, key = lambda x:x[1])
    for i in distances[1:6]:
        print(df.iloc[i[0]].Name)
        
recommend('Trigun')
        
```

```python
def recommend2(anime):
    index = df[df['Name'] ==anime].index[0]
    distances = sorted(list(enumerate(similarity[index])),reverse=True, key = lambda x:x[1])
    for i in distances[1:6]:
        print(df.iloc[i[0]].Name)
        print(df.iloc[i[0]].MAL_ID)
        
recommend2('Cowboy Bebop')      
```
